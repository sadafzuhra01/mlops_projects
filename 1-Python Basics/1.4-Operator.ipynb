#### 1- Arithematic oprator
These operator are used to perform basic mathematical operation like addition , substaction, multiplication,division.

```python
## +,-,*,/,//,%
num_1 = 40
num_2 = 5

sum =num_1+ num_2
sub = num_1 - num_2
mult = num_1 * num_2
div = num_1 / num_2
floor_div = num_1 // num_2 #floor division = near lowerest number without decimal
mod = num_1 % num_2 # modulus = remainder 

print (sum)
print (sub)
print (mult)
print (div)
print (floor_div)
print (mod)
```
