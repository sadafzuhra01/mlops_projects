```python
## if statement
age = 22
if age >= 18:
    print("you are eligible for voting")
```

```python
if age >= 18:
    print("you are eligible for voting")
    
else :
    print ("you are not eligible for voting ")   
```

```python
age = 22 
if age <= 12:
    print ("you are child")
 
elif age <= 19:
    print ("you are teenager")
        
elif age <=30:
    print ("you are adult")        

else:
    print ("you are old")   
```

```python
age = 21
if age <= 12:
    print ("you are child")
 
elif age <= 19:
    print ("you are teenager")
        
elif age <=30:
    print ("you are adult")        

elif age!=int:
    print ("invalid age")

else:
    print ("you are old")   
```

```python
## simple calculator 
num_1 = int(input("Enter a number!"))
num_2 =int(input("Enter a number!"))
operation =input("Enter a operation(+,-,*,/,//,%)")

if operation =='+':
    sum = num_1 + num_2
    print(sum)
elif operation == '-':
    diff = num_1 - num_2 
    print(diff)

elif operation == '*':
    mult =num_1 * num_2
    print(mult)

elif operation == '/':
    if num_2!=0:
        div  = num_1/ num_2 
    else:
        print("division by zero is not possible,chang the number.")
    print(div)

elif operation == '//':
    floor_div  = num_1// num_2 
    print(floor_div)

elif operation == '%':
    module  = num_1 % num_2 
    print(module)
   
else:
    print ("invalid syntax")
    
    

```

```python
## nested if -else 
year = int(input("Enter the year:"))

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print(f"The {year} is a leap year.")
        else:
            print(f"the {year} is not a leap year")
        print(f"the {year} is a leep year")            
    print(f"the {year} is a leep year")    
else:
    print(f"the {year} is not a leep year")    
```
