```python
## for loop 
age = [12,45,23,44,22,26,14]
for i in age:
    if i%2==0:
        print(i,end=' ')
```

```python
for i in range(100):
    if i%2!=0:
        print(i,end=' ')
```

```python
##tricky question 
sum =0 
for i in range(100):
    sum += i
    
print(sum)
```

```python
##psudocode 
num = 100 
fact = 1
count = 1
while count<=num:
    fact = fact * num
    count +=1
print(fact)   
```

```python
#while loop 
num = 1 
fact = 100
count = 1
while count<=1:
    fact = fact * count
    count-=1
print(fact)   
```

### loop control statment 


```python

##pass statment 
for i in range (100):
    pass
```

```python
##continue statement 
for i in range (10):
    if i%2==0:
        continue
    print(i,end=' ')
```

```python
##break stament
for i in range(50):
    if i == 38:
        break
    print(i,end=' ')
```

### Nested loops

```python
#nested loop
names = ['Sadaf','Noor','Eman','Ali'] 
for i in range(3):
    for j in names:
        print(f"{i}:{j}")
```

#### Nested loops with respect to biology

```python
# Find all combinations of DNA base pairs in a sequence list
sequences = ["ATCG", "GCTA", "TTAG"]
for seq in sequences:
    for base in seq:
        print(base,end=' ')
```

```python
##DNA sequence base count 
sequences = ["ATCG", "AATTGG", "CCGTTA"]

for seq in sequences:
    print(f"Sequence: {seq}")
    for base in "ATCG":
        count = seq.count(base)
        print(f"  {base}: {count}")
```

```python
## gene comparison 
genes = ["Gene1", "Gene2", "Gene3"]

for g1 in genes:
    for g2 in genes:
        if g1 != g2:
            print(f"Comparing {g1} with {g2}")
```

```python
expression = [
    [1.2, 3.4, 5.6],  # Gene 1
    [2.3, 4.5, 6.7],  # Gene 2
    [3.4, 5.6, 7.8]   # Gene 3
]

for gene_index, gene in enumerate(expression):
    print(f"Gene {gene_index + 1}:")
    for sample_index, value in enumerate(gene):
        print(f"  Sample {sample_index + 1}: {value}")
```
