```python
#nested tuple
tpl =((1,2,3,4,5),("Sadaf","eman","zainab"),(True ,False))
```

```python
# iterating over nested tuples
for sub_tpl in tpl :
    for i in sub_tpl:
        print(i , end =" ")#print each element in nested tuple
```

```python
for sub_tpl in tpl :
    for i in sub_tpl:
        if type(i)== str:
            print(i ,end =" ")
            
        else:
            continue    
```

```python
lst =[]
for sub_tpl in tpl :
    for i in sub_tpl:
        if type(i)== int:
            lst.append(i*22)
            
        else:
            continue    

print(lst)
```
