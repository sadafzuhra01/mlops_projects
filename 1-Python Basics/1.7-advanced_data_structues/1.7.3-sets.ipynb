### Sets in Python 
- Creating sets
- Accessing elements
- Methods and operation


```python
## sets 
##unorded collection of data ,unique elements 

new_set ={1,2,3,4,5,6,7}
print (type(new_set))
```

```python
# unique data point
new_set ={1,1,2,4,6,7,2,3,4,5,6,7}
print (new_set)
```

```python
lst =[1,1,2,4,6,7,2,3,4,5,6,7,4,3,2]
new_set = set(lst)
print (new_set)
```

```python
## method and operation
set_1 = {1,2,3,4,5}
set_1.add(6)
print(set_1) 
```

```python
set_1.remove(6)
print(set_1)
```

```python
set_2 = {6,7,8,9,0}
print(set_1.difference(set_2))## commen element 

print(set_1.intersection(set_2))

print(set_1.union(set_2))

```

```python
## symmetric difference 
##unique element in bith set 
print(set_1.symmetric_difference(set_2))
```
