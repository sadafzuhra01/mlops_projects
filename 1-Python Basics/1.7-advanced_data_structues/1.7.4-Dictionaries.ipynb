```python
## itrating over dictionaries 
for key,value in dict.items():
    print(f"{key}:{value}")
   
for key in dict.keys():
    print(key)
    
for value in dict.values():
    print(value)   
```

```python
## nested dictionaries
dict ={
    "personal details":{"name":"Sadaf","age":"22","email":"sadafzuhra788@gmail.com"},
    "contact details":{"phone number":"0345-1234567","address":"Karachi"},
}
```

```python
##itrating over nesting dictionaries
for key,value in dict.items():
    for sub_key,sub_value in value.items():
        print(f" {sub_key} : {sub_value}")
```

```python
## dictionaries comrehension
# {expresssion for item in itrable if condition}
square_dict = {x: x**2 for x in range(5)if x /2==0}

print(square_dict)
```

```python
## real world example 

# letter frequency 
letters = ["a","b","c","d","a","b","c","a","b"]
frequency = {}
for letter in letters :
    if letter in frequency :##membership test 
        frequency [letter] += 1
    else :
        frequency [letter] = 1
        
print(frequency)       
```
