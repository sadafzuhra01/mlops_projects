```python
## defining  a function
def even_or_odd(number):
    ## descriptionn 
    if number %2!=0:
        print(f"the {number} is an odd number.")
    else:
        print(f"the {number} is an even number.")
```

```python
## calling a function
num = 67
even_or_odd(num)
```

```python
## function with multiple argument 
def add(a,b):
    return a + b

def div (a,b):
    if b ==0:
        return "Error:Division by zero is not allowed"
    else:
        result = a/b
        return result
```

```python
print(add(2,9))
print(div(6,8))
```

```python
## function with default parameter
def greet(name = "Guest"):
    return f"Hey {name} , Welcome to UAF."
greet()
```

```python
## variable length Argument 
## positional and key word argument 
### positional ----> these are not in the key-value pair 
def print_function(*args):
    for arg in args:
        print(arg) 
        
print_function(1,2,3,4,5,6,7,8,"Sadaf","Noor")        

```

```python
def print_dict(**kwargs):
    for key ,value in kwargs.items():
        print(f"{key} :{value}")
  
print_dict(name="Sadaf",age = "23", inst= "UAF")        
```

```python
def print_gen(*args ,**kwargs):
    for arg in args:
        print(arg ,end =' ')
        
    for key,value in kwargs.items():
        print(f"{key}={value}", end = ' ')
        
print_gen(1,2,3,4,5,6,7,'Sadaf','Faisalabad', inst = 'UAF',)        
             
```

```python
### function with return key word only 
def square(a):
    result = a** 2
    return result

square(9)
```

## Map function 


```python
num1 = [1,2,3,4,5,6,9 ]

even = lambda x : x%2 == 0
list(map(even , num1))
```

```python
def cube(num):
    if num>5:
        return num**3
```

```python
num2= [4 , 5 ,7,6,9,8,2,1]
cube_result= list(map(cube ,num2))
print(cube_result)
```

## Filter function 

```python

def even (num):
    if num % 2 ==0 :
        return True
    
num2= [4 , 5 ,7,6,9,8,2,1]
filtered_function= list(filter(even , num2))    
print(filtered_function)
    
```
